## 1 - O Simulador "Desfazer"

```python
class EditorTexto:
    def __init__(self):
        self.texto = ""
        self.historico = []  # Pilha que guardará os estados anteriores do texto

    def salvar_estado(self):
        # Guarda a versão atual do texto na pilha antes de alterá-lo
        self.historico.append(self.texto)

    def digitar(self, texto_adicional: str):
        self.salvar_estado()
        self.texto = self.texto + texto_adicional
        print("Ação: Digitado '" + texto_adicional + "'")

    def editar(self, texto_completo: str):
        self.salvar_estado()
        self.texto = texto_completo
        print("Ação: Texto substituído por '" + texto_completo + "'")

    def deletar(self, quantidade_caracteres: int):
        self.salvar_estado()
        # O fatiamento [:-n] corta os últimos 'n' caracteres da string
        texto_removido = self.texto[-quantidade_caracteres:]
        self.texto = self.texto[:-quantidade_caracteres]
        print("Ação: Deletado '" + texto_removido + "'")

    def desfazer(self):
        if len(self.historico) == 0:
            print("Aviso: Nada para desfazer.")
            return

        # Remove o último estado salvo da pilha e substitui o texto atual
        self.texto = self.historico.pop()
        print("↺ Ação desfeita.")

    def exibir(self):
        print("---------------------------")
        print("Texto Atual: '" + self.texto + "'")
        print("---------------------------\n")


# --- Demonstração de Uso ---
editor = EditorTexto()

# 1. Testando a função Digitar
editor.digitar("Olá, ")
editor.digitar("mundo!")
editor.exibir()

# 2. Testando a função Deletar
editor.deletar(6)  # Remove "mundo!"
editor.exibir()

# 3. Testando a função Editar
editor.editar("Olá, estudante de Computação!")
editor.exibir()

# 4. Testando a função Desfazer (revertendo a Edição)
editor.desfazer()
editor.exibir()

# 5. Testando a função Desfazer novamente (revertendo o Delete)
editor.desfazer()
editor.exibir()
```

## Exercício 2 - Sistema de Impressão

```python
class SpoolerImpressora:
    def __init__(self):
        self.fila_impressao = []

    def adicionar_documento(self, nome_documento: str):
        self.fila_impressao.append(nome_documento)
        print("Documento '" + nome_documento + "' adicionado à fila de impressão.")

    def imprimir_proximo(self):
        if self.esta_vazia():
            print("Nenhum documento aguardando para ser impresso!")
            return None

        doc_impresso = self.fila_impressao.pop(0)
        print("Imprimindo: '" + doc_impresso + "'...")
        return doc_impresso

    def esta_vazia(self) -> bool:
        return len(self.fila_impressao) == 0

    def exibir_fila(self):
        if self.esta_vazia():
            print("Fila de impressão vazia.")
        else:
            print("\n--- Documentos na fila (Início -> Fim) ---")
            for pos, doc in enumerate(self.fila_impressao, start=1):
                print(str(pos) + "º da fila: " + doc)
            print("------------------------------------------")


# Demonstração
impressora = SpoolerImpressora()
impressora.adicionar_documento("Relatorio_Financeiro.pdf")
impressora.adicionar_documento("Trabalho_Faculdade.docx")
impressora.exibir_fila()

print("\n--- Processando Impressão ---")
impressora.imprimir_proximo()
impressora.exibir_fila()
```
