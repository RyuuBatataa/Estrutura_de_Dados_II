## 6. Referência versus cópia

### Referência

```python
a = [10, 20, 30]
b = a
b.append(40)

print(a)
print(b)
```

Resultado:

```text
[10, 20, 30, 40]
[10, 20, 30, 40]
```

### Cópia

```python
a = [10, 20, 30]
b = a.copy()
b.append(40)

print(a)
print(b)
```

Resultado:

```text
[10, 20, 30]
[10, 20, 30, 40]
```

## 13. Lista encadeada

```python
class ListaEncadeada:

    def __init__(self):
        self.inicio = None

    def adicionar(self, valor):
        novo = Node(valor)

        self.inicio = novo

    def exibir(self):
        atual = self.inicio

        while atual is not None:
            print(atual.valor)
            atual = atual.proximo

```

Responda:

1. Qual será a saída?
2. `a` e `b` representam o mesmo objeto?
3. Por quê?
4. O que aconteceria com `b = a.copy()`?

````python
a = [10, 20, 30]
b = a

b.append(40)

print("a:", a)
print("b:", b)
print("id(a):", id(a))
print("id(b):", id(b))
````

````python
a = [10, 20, 30]
b = b = a.copy()

b.append(40)

print("a:", a)
print("b:", b)
print("id(a):", id(a))
print("id(b):", id(b))
````

1. A saída de ambos resultará na mesma lista e no mesmo id.

2. Sim, representam o mesmo objeto e a saída da função id() comprova isso, pois ambas as variáveis compartilham o mesmo endereço na memória do computador.

3. Em Python, variáveis são como "etiquetas" coladas nos dados. Quando você faz a atribuição b = a, a linguagem não cria uma lista nova. Ela simplesmente pega a etiqueta b e cola no mesmo espaço de memória (objeto) onde a etiqueta a já estava.
Como ambas as etiquetas apontam para a mesma lista, qualquer alteração feita através de b (como o b.append(40)) refletirá quando você olhar para a. Chamamos isso de passagem por referência.

4. O valor de A é copiado para B pela função, mas não como o mesmo objeto. Por isso, quando o valor 40 é adicionado em B, não é copiado para A.


## 16. Exercício 2 — Construindo uma cadeia

Crie:

```python
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")
```

Conecte:

```text
A → B → C → None
```

Depois percorra a estrutura e produza:

```text
A
B
C
```


````python
# 1. Criando a estrutura base do Node
class Node:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None  # Por padrão, o próximo nó é vazio (None)

# 2. Criando os nós soltos (Cadeia)
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

# 3. Conectando os nós: A -> B -> C -> None
n1.proximo = n2  # O próximo depois do A é o B
n2.proximo = n3  # O próximo depois do B é o C
# O n3.proximo já é None por padrão, então finaliza a lista.

# 4. Percorrendo a estrutura e exibindo os valores
atual = n1  # Começamos pelo primeiro nó (A)

print("Percorrendo a lista encadeada:")
while atual is not None:
    print(atual.valor)
    atual = atual.proximo  # Pula para o próximo nó da corrente
````

## 18. Problema real — Sistema de atendimento de uma clínica

Uma clínica recebe pacientes durante o dia. Cada paciente possui:

- nome;
- idade;
- prioridade.

Exemplo:

```text
Ana — 32 anos — Normal
Bruno — 70 anos — Prioridade
Carlos — 45 anos — Normal
```

O sistema deve permitir:

1. adicionar paciente;
2. listar pacientes;
3. atender o primeiro paciente;
4. verificar se a fila está vazia;
5. informar a quantidade de pacientes.


````python
from dataclasses import dataclass

# Definimos a estrutura do Paciente
@dataclass
class Paciente:
    nome: str
    idade: int
    prioridade: str

# Criamos a classe para gerenciar a fila da clínica
class FilaClinica:
    def __init__(self):
        self.fila = [] # A fila começa vazia

    # 1. Adicionar paciente
    def adicionar(self, nome, idade, prioridade):
        novo_paciente = Paciente(nome, idade, prioridade)
        self.fila.append(novo_paciente) # Adiciona no final da lista
        print(f"✅ Paciente {nome} ({prioridade}) entrou na fila.")

    # 2. Listar pacientes
    def listar(self):
        if self.esta_vazia():
            print("📭 A fila está vazia no momento.")
            return

        print("\n--- 📋 Pacientes Aguardando ---")
        for i, p in enumerate(self.fila, start=1):
            print(f"{i}º da fila | {p.nome} - {p.idade} anos - {p.prioridade}")
        print("-------------------------------")

    # 3. Atender o primeiro paciente
    def atender(self):
        if self.esta_vazia():
            print("❌ Não há pacientes para atender.")
            return

        # O método pop(0) remove e retorna o elemento na posição 0 (o primeiro)
        paciente_atendido = self.fila.pop(0)
        print(f"\n🩺 ➡️ ATENDIMENTO: Chamando {paciente_atendido.nome} para o consultório.")

    # 4. Verificar se a fila está vazia
    def esta_vazia(self):
        return len(self.fila) == 0

    # 5. Informar a quantidade de pacientes
    def quantidade(self):
        return len(self.fila)

# ==========================================
# INTERFACE DO SISTEMA (MENU)
# ==========================================
sistema = FilaClinica()

while True:
    print("\n" + "="*30)
    print("🏥 SISTEMA DA CLÍNICA")
    print("1 - Adicionar paciente")
    print("2 - Listar pacientes")
    print("3 - Atender próximo paciente")
    print("4 - Verificar situação da fila")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ")
    print("="*30 + "\n")

    if opcao == '1':
        nome = input("Nome do paciente: ")
        idade = int(input("Idade: "))
        prioridade = input("Prioridade (Normal/Prioridade): ")
        sistema.adicionar(nome, idade, prioridade)

    elif opcao == '2':
        sistema.listar()

    elif opcao == '3':
        sistema.atender()

    elif opcao == '4':
        vazia = "Sim" if sistema.esta_vazia() else "Não"
        print(f"A fila está vazia? {vazia}")
        print(f"Total de pacientes aguardando: {sistema.quantidade()}")

    elif opcao == '0':
        print("Encerrando o sistema...")
        break

    else:
        print("⚠️ Opção inválida!")
````

````python
# Com Prioridade!

from dataclasses import dataclass

@dataclass
class Paciente:
    nome: str
    idade: int
    prioridade: str

class FilaClinica:
    def __init__(self):
        self.fila = []

    def adicionar(self, nome, idade, prioridade):
        novo_paciente = Paciente(nome, idade, prioridade)

        # Padroniza a string para evitar erros de digitação (maiuscula/minuscula)
        tipo_prioridade = prioridade.strip().lower()

        if tipo_prioridade == 'prioridade':
            # Procura a posição do primeiro paciente "Normal" para furar a fila
            posicao_insercao = 0
            for p in self.fila:
                if p.prioridade.strip().lower() == 'prioridade':
                    posicao_insercao += 1 # Pula quem já é prioridade
                else:
                    break # Achou o primeiro paciente "Normal", paramos aqui!

            # Insere exatamente na posição calculada, empurrando os normais para trás
            self.fila.insert(posicao_insercao, novo_paciente)
        else:
            # Se for normal, apenas entra no final da fila
            self.fila.append(novo_paciente)

        print(f"✅ Paciente {nome} ({prioridade}) entrou na fila.")

    def listar(self):
        if self.esta_vazia():
            print("📭 A fila está vazia no momento.")
            return

        print("\n--- 📋 Pacientes Aguardando ---")
        for i, p in enumerate(self.fila, start=1):
            print(f"{i}º da fila | {p.nome} - {p.idade} anos - {p.prioridade}")
        print("-------------------------------")

    def atender(self):
        if self.esta_vazia():
            print("❌ Não há pacientes para atender.")
            return

        paciente_atendido = self.fila.pop(0)
        print(f"\n🩺 ➡️ ATENDIMENTO: Chamando {paciente_atendido.nome} para o consultório.")

    def esta_vazia(self):
        return len(self.fila) == 0

    def quantidade(self):
        return len(self.fila)

# ==========================================
# INTERFACE DO SISTEMA (MENU)
# ==========================================
sistema = FilaClinica()

while True:
    print("\n" + "="*30)
    print("🏥 SISTEMA DA CLÍNICA")
    print("1 - Adicionar paciente")
    print("2 - Listar pacientes")
    print("3 - Atender próximo paciente")
    print("4 - Verificar situação da fila")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ")
    print("="*30 + "\n")

    if opcao == '1':
        nome = input("Nome do paciente: ")
        idade = int(input("Idade: "))
        prioridade = input("Prioridade (Normal/Prioridade): ")
        sistema.adicionar(nome, idade, prioridade)

    elif opcao == '2':
        sistema.listar()

    elif opcao == '3':
        sistema.atender()

    elif opcao == '4':
        vazia = "Sim" if sistema.esta_vazia() else "Não"
        print(f"A fila está vazia? {vazia}")
        print(f"Total de pacientes aguardando: {sistema.quantidade()}")

    elif opcao == '0':
        print("Encerrando o sistema...")
        break

    else:
        print("⚠️ Opção inválida!")
````
