```python
atividade = "ATIVIDADE 1 - Listas"
print(atividade)

# Cria a lista e solicita 10 números ao usuário
numeros = [int(input(f"Digite o {i+1}º número: ")) for i in range(10)]

# Cálculos básicos
soma = sum(numeros)
media = soma / len(numeros)

# Desafio: maior, menor e quantidade de pares
maior = max(numeros)
menor = min(numeros)
pares = sum(1 for n in numeros if n % 2 == 0)

print(f"\nLista completa: {numeros}")
print(f"Soma: {soma} | Média: {media}")
print(f"Maior: {maior} | Menor: {menor} | Pares: {pares}")
```

```python
atividade = "ATIVIDADE 2 - Matrizes"
print(atividade)

matriz = []

# Preenchendo a matriz 3x3
for i in range(3):
    linha = [int(input(f"Valor para [{i}][{j}]: ")) for j in range(3)]
    matriz.append(linha)

soma_total = 0
soma_diagonal = 0
maior_elemento = matriz[0][0]

print("\nMatriz:")
for i in range(3):
    for j in range(3):
        valor = matriz[i][j]
        print(f"{valor:^3}", end=" ")

        # Cálculos solicitados
        soma_total += valor
        if i == j:
            soma_diagonal += valor
        if valor > maior_elemento:
            maior_elemento = valor
    print() # Quebra de linha da matriz

print(f"\nSoma total: {soma_total}")
print(f"Soma da diagonal principal: {soma_diagonal}")
print(f"Maior elemento: {maior_elemento}")
```

```python
atividade = "ATIVIDADE 3 - Dicionários"
print(atividade)

# Registrando dados

livro = {
    "titulo": input("Título: "),
    "autor": input("Autor: "),
    "ano": int(input("Ano de publicação: ")),
    "preco": float(input("Preço: R$ "))
}

print("\nDados registrados:", livro)

# Alterando e adicionando dados
livro["preco"] = float(input("\nDigite o novo preço: R$ "))
livro["categoria"] = input("Digite a categoria do livro: ")

print("\nDados atualizados:", livro)
```

```python
atividade = "ATIVIDADE 4 - Dataclass"
print(atividade)

from dataclasses import dataclass

@dataclass
class Aluno:
    nome: str
    matricula: int
    nota1: float
    nota2: float

    def media(self):
        return (self.nota1 + self.nota2) / 2

# Atividade 5: Lista de objetos
alunos = []
print("Cadastre 5 alunos:")
for i in range(5):
    nome = input(f"\nNome do {i+1}º aluno: ")
    mat = int(input("Matrícula: "))
    n1 = float(input("Nota 1: "))
    n2 = float(input("Nota 2: "))
    alunos.append(Aluno(nome, mat, n1, n2))

print("\n--- Relatório de Aprovação ---")
maior_media = -1
melhor_aluno = None

for aluno in alunos:
    media_atual = aluno.media()
    if media_atual >= 7.0:
        print(f"Aprovado: {aluno.nome} - Média: {media_atual:.1f}")

    if media_atual > maior_media:
        maior_media = media_atual
        melhor_aluno = aluno

print(f"\nAluno com maior média: {melhor_aluno.nome} ({maior_media:.1f})")
```

```python
desafio = "DESAFIO - Estoque"
print(desafio)

from dataclasses import dataclass

@dataclass
class Produto:
    codigo: int
    nome: str
    preco: float
    quantidade: int

estoque = []

while True:
    print("\n===== SISTEMA DE ESTOQUE =====")
    print("1 - Cadastrar produto\n2 - Listar produtos\n3 - Buscar produto")
    print("4 - Valor total do estoque\n5 - Produto mais caro\n0 - Sair")

    escolha = input("Escolha: ")

    if escolha == '1':
        cod = int(input("Código: "))
        nome = input("Nome: ")
        preco = float(input("Preço: R$ "))
        qtd = int(input("Quantidade: "))
        estoque.append(Produto(cod, nome, preco, qtd))
        print("Produto cadastrado!")

    elif escolha == '2':
        for p in estoque:
            print(f"[{p.codigo}] {p.nome} - R${p.preco:.2f} (Qtd: {p.quantidade})")

    elif escolha == '3':
        busca = int(input("Código do produto: "))
        encontrado = next((p for p in estoque if p.codigo == busca), None)
        if encontrado:
            print(f"Encontrado: {encontrado.nome} (R${encontrado.preco:.2f})")
        else:
            print("Produto não encontrado.")

    elif escolha == '4':
        total = sum(p.preco * p.quantidade for p in estoque)
        print(f"Valor total do estoque: R${total:.2f}")

    elif escolha == '5':
        if estoque:
            mais_caro = max(estoque, key=lambda p: p.preco)
            print(f"Mais caro: {mais_caro.nome} (R${mais_caro.preco:.2f})")
        else:
            print("O estoque está vazio.")

    elif escolha == '0':
        print("Saindo...")
        break
    else:
        print("Opção inválida!")
```
